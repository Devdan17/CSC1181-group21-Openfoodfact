```python
import pandas as pd
```

```python
# Loading each member's cleaned parquet file

df1 = pd.read_parquet("Data_Cleaning/Devdan_Final_Output.parquet")
df2 = pd.read_parquet("Data_Cleaning/Devesh_Final_Output.parquet")
df3 = pd.read_parquet("Data_Cleaning/Rishabh_Final_Output.parquet")
df4 = pd.read_parquet("Data_Cleaning/Sahil_Final_Output.parquet")
```

```python
df1.head()
```

```python
df1.rename(columns={'URL': 'url'}, inplace=True)
```

```python
df2.head()
```

```python
df3.head()
```

```python
df4.head()
```

```python
# Merge them together using an 'outer' join on 'url'
# An 'outer' join ensures that if a product was kept by anyone, it stays in the final dataset.
final_df = df1.merge(df2, on="url", how="outer")
final_df = final_df.merge(df3, on="url", how="outer")
final_df = final_df.merge(df4, on="url", how="outer")
final_df.info()
```
